# ACE-Step 1.5: free song generator (Suno-style)
Makes full songs with vocals from lyrics + a style prompt. MIT license. Needs a GPU (runs under 4 GB VRAM with the 2B turbo model).

Runtime > Change runtime type > **T4 GPU** (Colab free) or Kaggle: Settings > Accelerator > **GPU T4 x2** and Internet **On**.

No-install option: https://huggingface.co/spaces/ACE-Step/Ace-Step-v1.5

```
!nvidia-smi
!pip -q install uv
!git clone https://github.com/ACE-Step/ACE-Step-1.5.git
%cd ACE-Step-1.5
!uv sync
```

Start the web UI. It prints a public **gradio.live** link: open it, paste lyrics (use [verse] / [chorus] tags) and a style like `raspy boom-bap rap, 92 BPM, C minor, war drums, brass`.

```
!uv run acestep --share
```

Download results from the UI, or zip the output folder:

```
!zip -r /content/ace_outputs.zip outputs || true
```
