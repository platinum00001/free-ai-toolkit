# Wan2.2 TI2V-5B: text/image to video
Apache-2.0. Turns a still (e.g. an Image Studio War Capital frame) into a ~5 s clip.

Runtime > Change runtime type > **T4 GPU** (Colab free) or Kaggle: Settings > Accelerator > **GPU T4 x2** and Internet **On**.
Wan's own README targets a 24 GB GPU for 720p; on a free 16 GB T4 use the offload flags below and expect slow runs. Easier free option: Wan / LTX Spaces listed in GPU_TOOLS.md.

```
!nvidia-smi
!git clone https://github.com/Wan-Video/Wan2.2.git
%cd Wan2.2
!pip -q install -r requirements.txt "huggingface_hub[cli]"
!huggingface-cli download Wan-AI/Wan2.2-TI2V-5B --local-dir ./Wan2.2-TI2V-5B
```

Upload your still to `/content/input.png` (Files panel), edit the prompt, run:

```
PROMPT = "Armored arctic convoy rolls across an ice ridge at dusk, snow blowing, cyan light bar glowing, slow cinematic push-in"
!python generate.py --task ti2v-5B --size 1280*704 --ckpt_dir ./Wan2.2-TI2V-5B --offload_model True --convert_model_dtype --t5_cpu --image /content/input.png --prompt "$PROMPT"
```
