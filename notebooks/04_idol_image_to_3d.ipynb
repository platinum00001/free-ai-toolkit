# Idol: one image to a 3D Gaussian splat (.ply)
Apache-2.0, ~7.2 GB VRAM, fits a free T4. Good for Character Builder art and War Capital vehicles.

Runtime > Change runtime type > **T4 GPU** (Colab free) or Kaggle: Settings > Accelerator > **GPU T4 x2** and Internet **On**.

No-install option: https://huggingface.co/spaces/speridlabs/idol . For textured GLB meshes use https://huggingface.co/spaces/microsoft/TRELLIS.2 (needs 24 GB locally, so use the Space).

```
!git clone https://github.com/speridlabs/idol-sdk.git
%cd idol-sdk
!pip -q install -e .
```

Upload your picture to `/content/input.png`, then:

```
from idol import IdolPipeline
pipeline = IdolPipeline.from_pretrained('speridlabs/idol')
out = pipeline('/content/input.png', mode='real', seed=42, num_gaussians=1_048_576)
out.gaussian.save_ply('/content/model.ply')
print('saved /content/model.ply (view at https://superspl.at/editor or load in three.js with a splat viewer)')
```
