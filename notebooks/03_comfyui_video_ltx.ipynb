Runtime > Change runtime type > **T4 GPU** (Colab free) or Kaggle: Settings > Accelerator > **GPU T4 x2** and Internet **On**.

```
!git clone https://github.com/Comfy-Org/ComfyUI.git
%cd ComfyUI
!pip -q install -r requirements.txt
!git clone https://github.com/Comfy-Org/ComfyUI-Manager.git custom_nodes/ComfyUI-Manager
!wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64 -O /content/cloudflared && chmod +x /content/cloudflared
```
